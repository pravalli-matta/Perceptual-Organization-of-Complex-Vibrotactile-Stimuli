```python
def update_matrix(
    matrix,
    stimA,
    stimB,
    response
):
    
    i = int(stimA[1:]) - 1
    j = int(stimB[1:]) - 1

    matrix[i,j] = response
    matrix[j,i] = response

    return matrix
```

```python
from scipy.io import wavfile
import sounddevice as sd

def play_wav(filepath):

    fs, data = wavfile.read(filepath)

    sd.play(
        data,
        samplerate=fs,
        blocking=True
    )
```

```python
def collect_response(root):

    response = []
    rt = []

    start_time = time.time()

    def on_key(event):

        if event.keysym in ["1", "KP_1"]:

            if len(response) == 0:

                response.append(1)

                rt.append(
                    (time.time() - start_time) * 1000
                )

        elif event.keysym in ["3", "KP_3"]:

            if len(response) == 0:

                response.append(0)

                rt.append(
                    (time.time() - start_time) * 1000
                )

    root.bind("<Key>", on_key)

    while time.time() - start_time < 2.0:

        root.update()

    root.unbind("<Key>")

    if len(response) == 0:

        return None, np.nan

    return response[0], rt[0]
```

```python
def run_trial(
    root,
    stimA,
    stimB,
    wav_lookup
):

    # -------------------
    # Fixation
    # -------------------

    clear_widgets(root)

    fixation = tk.Label(
        root,
        text="+",
        font=("Arial", 36),
        fg="white",
        bg="black"
    )

    fixation.pack(expand=True)

    root.update()

    time.sleep(0.3)

    fixation.destroy()

    # -------------------
    # Stimulus A
    # -------------------

    play_wav(
        wav_lookup[stimA]
    )

    # -------------------
    # ISI
    # -------------------

    time.sleep(0.150)

    # -------------------
    # Stimulus B
    # -------------------

    play_wav(
        wav_lookup[stimB]
    )

    # -------------------
    # Blank Response Screen
    # -------------------

    clear_widgets(root)

    root.update()
    response_label = tk.Label(
      root,
      text="RESPOND",
      font=("Arial", 24),
      fg="white",
      bg="black"
)

    response_label.pack(expand=True)

    root.update()
    response, rt = collect_response(root)

    # -------------------
    # ITI
    # -------------------

    time.sleep(0.150)

    return response, rt
```

```python
def clear_widgets(root):

    for widget in root.winfo_children():

        widget.destroy()
```

```python
import tkinter as tk

root = tk.Tk()
root.focus_set()

root.configure(bg="black")

root.geometry("800x600")

response, rt = run_trial(
    root,
    "S01",
    "S05",
    wav_lookup
)

print(response)
print(rt)

root.destroy()
```

```python
import tkinter as tk
from tkinter import simpledialog

def get_participant_id():

    root = tk.Tk()
    root.withdraw()

    participant_id = simpledialog.askstring(
        "Participant ID",
        "Enter Participant ID:"
    )

    root.destroy()

    return participant_id
```

```python
from pathlib import Path

results_dir = Path("../results")

results_dir.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
def block_break(root, block_num):

    clear_widgets(root)

    label = tk.Label(
        root,
        text=f"End of Block {block_num}\n\nTake a short break\n\nPress SPACE to continue",
        font=("Arial",24),
        fg="white",
        bg="black"
    )

    label.pack(expand=True)

    waiting = [True]

    def continue_exp(event):

        if event.keysym == "space":

            waiting[0] = False

    root.bind("<Key>", continue_exp)

    while waiting[0]:

        root.update()

    root.unbind("<Key>")
```

```python
def end_screen(root):

    clear_widgets(root)

    label = tk.Label(
        root,
        text="Experiment Complete\n\nThank You",
        font=("Arial",28),
        fg="white",
        bg="black"
    )

    label.pack(expand=True)

    root.update()

    time.sleep(3)
```

```python
participant_id = get_participant_id()

trial_results = []

participant_matrix = np.full(
    (27,27),
    np.nan
)

root = tk.Tk()

root.configure(bg="black")

root.geometry("800x600")
```

```python
for block_num, block in enumerate(blocks, start=1):

    for trial_num, row in block.iterrows():

        stimA = row["stimulus_A"]
        stimB = row["stimulus_B"]

        response, rt = run_trial(
            root,
            stimA,
            stimB,
            wav_lookup
        )

        update_matrix(
            participant_matrix,
            stimA,
            stimB,
            response
        )

        trial_results.append({
            "participant": participant_id,
            "block": block_num,
            "trial": trial_num + 1,
            "stimulus_A": stimA,
            "stimulus_B": stimB,
            "response": response,
            "rt_ms": rt
        })

    if block_num < 5:

        block_break(
            root,
            block_num
        )
```

```python
trial_df = pd.DataFrame(
    trial_results
)

trial_df.to_csv(
    results_dir /
    f"participant_{participant_id}_trials.csv",
    index=False
)
```

```python
matrix_df = pd.DataFrame(
    participant_matrix
)

matrix_df.to_csv(
    results_dir /
    f"participant_{participant_id}_matrix.csv",
    index=False
)
```
