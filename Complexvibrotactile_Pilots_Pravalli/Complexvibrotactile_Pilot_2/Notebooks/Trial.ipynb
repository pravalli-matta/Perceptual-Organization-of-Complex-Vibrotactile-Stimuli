```python
from pathlib import Path
from scipy.io.wavfile import read
import sounddevice as sd

BASE_DIR = Path.cwd().parent    # Since you're inside Notebooks
STIMULUS_FOLDER = BASE_DIR / "stimuli" / "waveform_27"

# Practice stimuli
trial1 = STIMULUS_FOLDER / "S01_sine_IFI50_AM30.wav"
trial2 = STIMULUS_FOLDER / "S27_saw_IFI150_AM90.wav"

def play_trial(filepath):
    fs, signal = read(filepath)

    signal = signal.astype(float)

    if signal.max() > 1:
        signal /= 32767

    sd.stop()
    sd.play(signal, fs)
    sd.wait()

print("Playing Practice Stimulus 1...")
play_trial(trial1)

input("Press Enter to play Practice Stimulus 2...")

print("Playing Practice Stimulus 2...")
play_trial(trial2)

print("Practice complete.")
```
